# 사칙연산
- 두 자연수 A와 B가 주어진다. 이때, A+B, A-B, A*B, A/B(몫), A%B(나머지)를 출력하는 프로그램을 작성하시오. 

```python
a, b = map(int, input().split())
def operate(a, b):
    print(a + b)
    print(a - b)
    print(a * b)
    print(a // b)
    print(a % b)
operate(a, b)
```

# 나머지
- (A+B)%C는 ((A%C) + (B%C))%C 와 같을까?
- (A×B)%C는 ((A%C) × (B%C))%C 와 같을까?
- 세 수 A, B, C가 주어졌을 때, 위의 네 가지 값을 구하는 프로그램을 작성하시오.

```python
a, b, c = map(int, input().split())
def remain(a, b, c):
    i = (a+b)%c
    j = ((a%c)+(b%c))%c
    k = (a*b)%c
    l = ((a%c)*(b%c))%c
    print(i)
    print(j)
    print(k)
    print(l)
remain(a, b, c)
```
